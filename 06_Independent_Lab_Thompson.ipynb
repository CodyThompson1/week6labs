Welcome to my notebook for the Grade Entry System! This program simulates a student management system where I can manage student information, query GPA metrics, and assign academic violations. The system supports various functionalities through a menu-driven interface.


```python
import random as rand  # Importing the random library for random number generation

# Initialize lists to store student data, academic violations, and GPA metrics
students = [
    {"first_name": "Carmen", "last_name": "Berzatto", "major": "Business", "gpa": 3.7},
    {"first_name": "Sydney", "last_name": "Adamu", "major": "Engineering", "gpa": 3.9},
    {"first_name": "Neil", "last_name": "Fak", "major": "Mathematics", "gpa": 3.5},
    {"first_name": "Marcus", "last_name": "Brooks", "major": "History", "gpa": 3.6}
]

violations = ["", "", "", ""]  # List to store academic violations, initially empty for each student
gpas = [3.7, 3.9, 3.5, 3.6]  # List to store GPA of all students for GPA metrics calculations

# Function to display the menu
def display_menu():
    print("\nWelcome to the Grade Entry System!")
    print("1. Student Information Query")
    print("2. Student Information Entry")
    print("3. Query GPA Metrics")
    print("4. Academic Violations")
    print("5. Quit Program")

# Function to query student information (with violations)
def query_student_info():
    print("\nStudent Information:")
    for i, student in enumerate(students):
        violation = violations[i] if violations[i] else "No violation"
        print(f"{student['first_name']} {student['last_name']} | Major: {student['major']} | GPA: {student['gpa']} | Violation: {violation}")

# Function to add new student information
def add_student():
    first_name = input("\nEnter the student's first name: ")
    last_name = input("Enter the student's last name: ")
    major = input("Enter the student's major: ")
    gpa = float(input("Enter the student's GPA: "))
    # Add the new student to the system
    students.append({"first_name": first_name, "last_name": last_name, "major": major, "gpa": gpa})
    gpas.append(gpa)  # Add the GPA to the list for GPA metrics calculation
    violations.append("")  # Add an empty violation entry for the new student

# Function to query GPA metrics (max, min, and average)
def query_gpa_metrics():
    max_gpa = max(gpas)
    min_gpa = min(gpas)
    avg_gpa = sum(gpas) / len(gpas)
    print(f"\nMax GPA: {max_gpa}")
    print(f"Min GPA: {min_gpa}")
    print(f"Average GPA: {avg_gpa:.2f}")

# Function to assign academic violations (only one per student)
def assign_violations():
    print("\nSelect a student to assign a violation:")
    for i, student in enumerate(students):
        print(f"{i+1}. {student['first_name']} {student['last_name']}")
    
    try:
        student_choice = int(input("\nEnter the number of the student (1-{}): ".format(len(students))))
        if student_choice < 1 or student_choice > len(students):
            print("Invalid choice. Please try again.")
            return
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        return

    # Check if the student already has a violation
    if violations[student_choice - 1] != "":
        print(f"\n{students[student_choice - 1]['first_name']} {students[student_choice - 1]['last_name']} already has a violation: {violations[student_choice - 1]}")
        return

    # List of possible violations
    violations_list = [
        "Copying homework",
        "Cheating on exam",
        "Plagiarism",
        "Cheating on quiz",
        "Writing paper for another student"
    ]
    
    print("\nSelect an academic violation:")
    for i, violation in enumerate(violations_list):
        print(f"{i+1}. {violation}")
    
    try:
        violation_choice = int(input("\nEnter the number of the violation (1-5): "))
        if violation_choice < 1 or violation_choice > len(violations_list):
            print("Invalid violation choice. Please try again.")
            return
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        return
    
    # Assign violation to the selected student
    violations[student_choice - 1] = violations_list[violation_choice - 1]
    print(f"\nViolation '{violations_list[violation_choice - 1]}' assigned to {students[student_choice - 1]['first_name']} {students[student_choice - 1]['last_name']}.")

# Main program loop
while True:
    display_menu()  # Display the menu
    choice = input("\nSelect an option (1-5): ")

    if choice == '1':
        query_student_info()
    elif choice == '2':
        add_student()
    elif choice == '3':
        query_gpa_metrics()
    elif choice == '4':
        assign_violations()
    elif choice == '5':
        print("\nThank you for using the Grade Entry System!")
        break
    else:
        print("\nInvalid option. Please try again.")
```
