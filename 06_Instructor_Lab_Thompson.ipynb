Welcome to my notebook for the Magic Crystal Ball program! In this notebook, I will simulate a magic crystal ball that answers user questions and tracks the responses. The program will randomly select answers from a list of predefined responses, and will also keep track of how much the user owes for their services.


```python
# import random as rand  # Importing the random library for random number generation

# List of responses (8 predefined responses + 3 custom ones)
responses = [
    "It is certain",
    "It is decidedly so",
    "That question requires greater monetary compensation",
    "Reply is hazy; try again, with more money",
    "Concentrate and ask again",
    "My reply is no",
    "Outlook is not so good",
    "Very doubtful",
    "Yes, but only if Duke wins",
    "Are you in an earthquake because the answer is shaky",
    "Just because I like you.... YES!"
]

# Initialize variables
questions = []  # List to store the user's questions
answers = []  # List to store the answers corresponding to the questions
total_cost = 0  # Initialize the total cost
question_count = 0  # Count the number of questions asked
asking_question = True  # Boolean to check if a question has been asked

# Start the program
print("Welcome to the all-seeing computer!\n")

while True:
    # Step 4: Display the menu options
    print("1. Ask a question")
    print("2. Review questions and responses")
    print("3. Complete services and pay")

    # Get the user's choice from the menu
    menu_choice = input("\nPlease select an option (1-3): ")

    if menu_choice == '1':
        if asking_question:
            # Ask the user to enter a question
            question = input("\nAsk what you seek to know...\n")
            # Randomly select a response from the list
            response = rand.choice(responses)
            # Store the question and response
            questions.append(question)
            answers.append(response)
            total_cost += 30  # Add $30 for each question asked
            question_count += 1  # Increment the question count
            print(f"\n{response}\n")
            asking_question = False  # Set asking_question to False after first question
        else:
            # If a question was already asked, prompt for another question
            question = input("\nAsk what you seek to know...\n")
            response = rand.choice(responses)  # Randomly select another response
            questions.append(question)  # Add the new question
            answers.append(response)  # Add the new response
            total_cost += 30  # Add $30 for this additional question
            question_count += 1  # Increment the question count
            print(f"\n{response}\n")
        
    elif menu_choice == '2':
        # Review the questions and answers
        print("\nYour questions and responses:")
        for i in range(question_count):
            print(f"Question: \"{questions[i]}\"")
            print(f"Response: \"{answers[i]}\"\n")

    elif menu_choice == '3':
        # Complete services and show the final amount owed
        print("\nThank you for visiting the all-seeing computer.")
        print(f"\nYou asked {question_count} questions.")
        print(f"You owe ${total_cost:.2f} for today's services.")
        break  # End the program once the user is done

    else:
        # Handle invalid input
        print("\nThat is not an option. Please try again.")
```
